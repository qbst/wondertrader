# 模块概述

WtUftRunner 是 WonderTrader UFT（Ultra Fast Trading，极速交易）策略运行器模块，负责协调和管理UFT策略系统的各个组件。主要包括：
- UFT策略系统初始化和配置管理
- UFT引擎管理
- 交易通道和行情通道管理
- UFT策略生命周期管理
- 共享内存域管理
- 数据管理和事件通知

1. **运行器层**（WtUftRunner）：
   - WtUftRunner：UFT策略运行器类，统一管理UFT策略系统的所有组件
   - main：主程序入口，提供命令行参数解析和程序启动
   - 作为UFT策略系统的主入口，负责初始化、配置和运行整个UFT交易系统

2. **引擎层**（WtUftEngine）：
   - WtUftEngine：UFT策略引擎，负责策略上下文管理、数据订阅、数据分发、时间管理等
   - 是整个UFT框架的控制中枢，管理策略的运行环境

3. **策略管理层**（UftStrategyMgr）：
   - UftStrategyMgr：UFT策略管理器，负责加载UFT策略工厂，创建和管理UFT策略实例
   - 负责策略的动态加载和生命周期管理

4. **适配器层**（TraderAdapterMgr + ParserAdapterMgr）：
   - TraderAdapterMgr：交易适配器管理器，管理多个交易通道，适配不同的交易接口
   - ParserAdapterMgr：解析器适配器管理器，管理多个行情解析通道，适配不同的行情数据源
   - 提供统一的交易和行情接口

5. **数据管理层**（WtUftDtMgr + WTSBaseDataMgr）：
   - WtUftDtMgr：UFT数据管理器，负责实时和历史数据管理，提供数据查询接口
   - WTSBaseDataMgr：基础数据管理器，管理合约、商品、交易时段等基础信息

6. **工具支持层**（EventNotifier + ActionPolicyMgr + ShareManager）：
   - EventNotifier：事件通知器，负责事件通知和回调
   - ActionPolicyMgr：动作策略管理器，管理交易动作的执行策略
   - ShareManager：共享内存管理器，管理策略参数的共享和持久化，支持跨进程参数监控

## 系统初始化与配置

### 初始化运行器 init
```cpp
/**
 * @brief 初始化运行器实现
 * @param filename 日志配置文件路径
 * 
 * 初始化日志系统，设置WonderTrader的工作目录。
 * 该函数应在创建实例后立即调用，在config()之前调用。
 */
void WtUftRunner::init(const std::string& filename)
{
	WTSLogger::init(filename.c_str());  // 初始化日志系统，从文件加载日志配置

	WtHelper::setInstDir(getBinDir());  // 设置WonderTrader的工作目录（可执行文件所在目录）
}
```

### 配置运行器 config
```cpp
/**
 * @brief 配置运行器实现
 * @param filename 主配置文件路径
 * @return 配置成功返回true，失败返回false
 * 
 * 加载主配置文件，初始化各个组件。
 * 配置流程：
 * 1. 加载配置文件
 * 2. 加载基础数据（交易时段、商品、合约、节假日）
 * 3. 初始化引擎
 * 4. 初始化数据管理器
 * 5. 初始化共享内存域（如果配置）
 * 6. 初始化动作策略管理器
 * 7. 初始化解析器适配器
 * 8. 初始化交易适配器
 * 9. 初始化UFT策略
 */
bool WtUftRunner::config(const std::string& filename)
```

## 运行运行器 run
```cpp
/**
 * @brief 运行运行器实现
 * @param bAsync 是否异步运行（默认false，当前版本未使用）
 * 
 * 启动引擎、解析器、交易适配器，并进入主循环监控运行状态。
 * 如果配置了共享内存域，会启动参数监控。
 * 主循环会一直运行直到_to_exit标志为true（由信号钩子设置）。
 */
void WtUftRunner::run(bool bAsync /* = false */)
{
	try  // 异常处理块
	{
		_uft_engine.run();  // 启动UFT引擎（初始化策略上下文，创建实时ticker等）

		_parsers.run();  // 启动解析器适配器（启动行情接收线程）

		_traders.run();  // 启动交易适配器（启动交易通道连接）

		ShareManager::self().start_watching(2);  // 启动共享内存参数监控（监控间隔2微秒，0表示无限循环检查）

		while(!_to_exit)  // 主循环，直到退出标志为true
		{
			std::this_thread::sleep_for(std::chrono::milliseconds(10));  // 休眠10毫秒，避免CPU占用过高
		}
	}
	catch (...)  // 捕获所有异常
	{
		print_stack_trace([](const char* message) {  // 打印堆栈跟踪信息
			WTSLogger::error(message);  // 将错误信息记录到日志系统
		});
	}
}
```

## WtUftRunner/main.cpp 的 main 函数
WonderTrader UFT（Ultra Fast Trading）策略运行器的主入口，用于启动和管理 UFT 策略的实盘运行。

1. 崩溃转储初始化（Windows）
   - 启用 MiniDump，模块名 "WtUftRunner.exe"，完整转储
   - 用于程序崩溃时生成转储文件
2. 命令行参数解析
   - `-c, --config`：配置文件路径（默认 `./config.yaml`）
   - `-l, --logcfg`：日志配置文件路径（默认 `./logcfg.yaml`）
   - `-h, --help`：显示帮助并退出
3. 日志系统初始化
   - 使用日志配置文件初始化日志系统
   - 若未指定，使用默认路径 `./logcfg.yaml`
4. 创建并初始化 UFT 运行器
   - 创建 `WtUftRunner` 实例
   - 调用 `init()` 初始化日志系统
5. 加载主配置
   - 使用配置文件初始化各组件
   - 若未指定，使用默认路径 `./config.yaml`
   - 调用 `config()` 加载配置并初始化组件，包括：
     - 加载基础数据（交易时段、商品、合约、节假日）
     - 初始化 UFT 引擎
     - 初始化数据管理器
     - 初始化共享内存域（如果配置）
     - 初始化动作策略管理器
     - 初始化解析器适配器
     - 初始化交易适配器
     - 初始化 UFT 策略
6. 启动运行（同步模式）
   - 调用 `runner.run(false)`，阻塞主线程
   - 运行期间：
     - 启动引擎、解析器、交易适配器
     - 如果配置了共享内存域，启动参数监控
     - 主循环持续运行直到 `_to_exit` 标志为 true
7. 程序退出
   - 返回 0
