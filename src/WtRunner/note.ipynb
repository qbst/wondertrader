# 模块概述

WtRunner 是 WonderTrader 标准策略运行器模块，负责统一管理和协调各个组件的生命周期，是 WonderTrader 系统的核心运行器。主要包括：
- 系统初始化和配置管理
- 多引擎支持（CTA引擎、HFT引擎、选股引擎）
- 策略生命周期管理
- 交易通道和行情通道管理
- 执行器管理
- 数据管理和事件通知

1. **运行器层**（WtRunner）：
   - WtRunner：核心运行器类，统一管理所有组件的生命周期
   - main：主程序入口，提供命令行参数解析和程序启动
   - 作为 WonderTrader 系统的主入口，负责初始化、配置和运行整个交易系统

2. **引擎层**（WtCtaEngine/WtHftEngine/WtSelEngine）：
   - WtCtaEngine：CTA策略引擎，管理CTA策略的运行环境
   - WtHftEngine：HFT策略引擎，管理高频交易策略的运行环境
   - WtSelEngine：选股引擎，管理选股策略的运行环境
   - 根据配置自动选择对应的引擎类型

3. **策略管理层**（CtaStrategyMgr/HftStrategyMgr/SelStrategyMgr）：
   - CtaStrategyMgr：CTA策略管理器，加载CTA策略工厂，创建和管理CTA策略实例
   - HftStrategyMgr：HFT策略管理器，加载HFT策略工厂，创建和管理HFT策略实例
   - SelStrategyMgr：选股策略管理器，加载选股策略工厂，创建和管理选股策略实例
   - 负责策略的动态加载和生命周期管理

4. **适配器层**（TraderAdapterMgr + ParserAdapterMgr）：
   - TraderAdapterMgr：交易适配器管理器，管理多个交易通道，适配不同的交易接口
   - ParserAdapterMgr：解析器适配器管理器，管理多个行情通道，适配不同的行情数据源
   - 提供统一的交易和行情接口

5. **执行器层**（WtExecuterFactory）：
   - WtExecuterFactory：执行器工厂，创建和管理执行器实例（本地执行器、差分执行器、分布式执行器等）
   - 负责订单的执行和路由管理
   - 仅用于CTA和选股策略，HFT策略不使用执行器

6. **数据管理层**（WtDtMgr + WTSBaseDataMgr + WTSHotMgr）：
   - WtDtMgr：数据管理器，管理实时行情数据和K线数据
   - WTSBaseDataMgr：基础数据管理器，管理商品、合约、交易时段等基础数据
   - WTSHotMgr：热点合约管理器，管理主力合约、次主力合约等

7. **工具支持层**（EventNotifier + ActionPolicyMgr）：
   - EventNotifier：事件通知器，处理系统事件通知和回调
   - ActionPolicyMgr：开平策略管理器，管理开仓和平仓策略的执行规则

# 策略运行器类 WtRunner.h/cpp

## 成员
- **配置对象**
  - `WTSVariant* _config`：配置对象指针，存储加载的配置信息

- **适配器管理器**
  - `TraderAdapterMgr _traders`：交易适配器管理器，管理多个交易通道
  - `ParserAdapterMgr _parsers`：解析器适配器管理器，管理多个行情通道

- **执行器工厂**
  - `WtExecuterFactory _exe_factory`：执行器工厂，创建和管理执行器实例（本地执行器、差分执行器、分布式执行器等）

- **交易引擎实例**
  - `WtCtaEngine _cta_engine`：CTA引擎实例，用于CTA策略运行
  - `WtHftEngine _hft_engine`：HFT引擎实例，用于HFT策略运行
  - `WtSelEngine _sel_engine`：选股引擎实例，用于选股策略运行
  - `WtEngine* _engine`：当前使用的引擎指针，指向_cta_engine、_hft_engine或_sel_engine之一（根据配置选择）

- **数据存储与管理**
  - `WtDataStorage* _data_store`：数据存储对象指针，用于数据持久化（当前未使用）
  - `WtDtMgr _data_mgr`：数据管理器，管理行情数据和K线数据

- **基础数据管理器**
  - `WTSBaseDataMgr _bd_mgr`：基础数据管理器，管理商品、合约、交易时段等基础数据
  - `WTSHotMgr _hot_mgr`：热点合约管理器，管理主力合约、次主力合约等

- **事件通知器**
  - `EventNotifier _notifier`：事件通知器，处理系统事件通知

- **策略管理器**
  - `CtaStrategyMgr _cta_stra_mgr`：CTA策略管理器，管理CTA策略工厂和策略实例
  - `HftStrategyMgr _hft_stra_mgr`：HFT策略管理器，管理HFT策略工厂和策略实例
  - `SelStrategyMgr _sel_stra_mgr`：选股策略管理器，管理选股策略工厂和策略实例

- **开平策略管理器**
  - `ActionPolicyMgr _act_policy`：开平策略管理器，管理开仓和平仓策略

- **状态标志**
  - `bool _is_hft`：是否为高频引擎标志，true表示使用HFT引擎，false表示使用CTA或选股引擎
  - `bool _is_sel`：是否为选股引擎标志，true表示使用选股引擎，false表示使用CTA或HFT引擎
  - `bool _to_exit`：退出标志，true表示需要退出，false表示继续运行

## 系统初始化与配置

### 初始化策略运行器 init
```cpp
/**
 * @brief 初始化策略运行器
 * 
 * 初始化日志系统和运行环境。
 * 
 * @param filename 日志配置文件路径
 * 
 * 初始化流程：
 * 1. 初始化日志系统（从文件加载配置）
 * 2. 设置安装目录路径（用于查找动态库和配置文件）
 * 3. 检查日志配置文件是否存在（如果不存在则记录警告）
 */
void WtRunner::init(const std::string& filename)
{
	WTSLogger::init(filename.c_str());  // 初始化日志系统，从文件加载日志配置

	WtHelper::setInstDir(getBinDir());  // 设置安装目录路径，用于查找动态库和配置文件

	if(!StdFile::exists(filename.c_str()))  // 如果日志配置文件不存在
	{
		WTSLogger::warn("logging configure {} not exists", filename);  // 记录警告日志
	}
}
```

### 配置策略运行器 config
```cpp
/**
 * @brief 配置策略运行器
 * 
 * 从配置文件加载配置并初始化各组件。
 * 
 * @param filename 配置文件路径
 * @return 配置成功返回true，失败返回false
 * 
 * 配置流程：
 * 1. 加载主配置文件
 * 2. 加载基础数据文件（交易时段、商品、合约、节假日、热点合约、次主力合约、自定义规则等）
 * 3. 标记合约的热点标志（主力合约、次主力合约）
 * 4. 初始化交易引擎（根据配置选择CTA、HFT或选股引擎）
 * 5. 初始化数据管理器
 * 6. 初始化开平策略
 * 7. 初始化行情通道（解析器）
 * 8. 初始化交易通道（交易适配器）
 * 9. 初始化事件通知器
 * 10. 如果不是高频引擎，初始化执行器和路由规则
 * 11. 初始化策略（CTA策略或HFT策略）
 */
bool WtRunner::config(const std::string& filename)
```

## 运行策略运行器 run
```cpp
/**
 * @brief 运行策略运行器
 * 
 * 启动各组件运行。
 * 
 * @param bAsync 是否异步运行，true表示异步运行（不阻塞），false表示同步运行（阻塞）
 */
void WtRunner::run(bool bAsync /* = false */)
{
	try  // 捕获异常
	{
		_parsers.run();  // 启动解析器适配器管理器（行情通道）
		_traders.run();  // 启动交易适配器管理器（交易通道）

		_engine->run();  // 启动交易引擎，运行策略逻辑

		if(!bAsync)  // 如果同步运行
		{
			while(!_to_exit)  // 循环等待退出信号
			{
				std::this_thread::sleep_for(std::chrono::milliseconds(10));  // 每次睡眠10毫秒，避免CPU占用过高
			}
		}
	}
	catch (...)  // 捕获所有异常
	{
		print_stack_trace([](const char* message) {  // 打印堆栈跟踪信息，使用lambda表达式作为回调函数
			WTSLogger::error(message);  // 将堆栈跟踪信息记录到日志
		});
	}
}
```

## ILogHandler 接口实现

### 处理日志追加事件 handleLogAppend
```cpp
/**
 * @brief 处理日志追加事件（ILogHandler接口实现）
 * 
 * 当系统产生新的日志时调用，将日志转发给事件通知器。
 * 
 * @param ll 日志级别（WTSLogLevel枚举值）
 * @param msg 日志消息内容
 */
void WtRunner::handleLogAppend(WTSLogLevel ll, const char* msg)
{
	_notifier.notify_log(LOG_TAGS[ll - 100], msg);  // 通过事件通知器通知日志事件
	// 参数说明：
	// - LOG_TAGS[ll - 100]: 日志标签字符串（将日志级别转换为标签）
	// - msg: 日志消息内容
}
```

# main.cpp/main
WonderTrader 标准策略运行器的主入口，用于启动和管理 CTA/HFT/选股策略的实盘运行。

1. 崩溃转储初始化（Windows）
   - 启用 MiniDump，模块名 "WtRunner.exe"，完整转储
   - 用于程序崩溃时生成转储文件
2. 命令行参数解析
   - `-c, --config`：配置文件路径（默认 `./config.yaml`）
   - `-l, --logcfg`：日志配置文件路径（默认 `./logcfg.yaml`）
   - `-h, --help`：显示帮助并退出
3. 日志系统初始化
   - 使用日志配置文件初始化日志系统
   - 若未指定，使用默认路径 `./logcfg.yaml`
4. 创建并初始化运行器
   - 创建 `WtRunner` 实例 `runner`
   - 调用 `runner.init()` 初始化日志系统
5. 加载主配置
   - 使用配置文件初始化各组件
   - 若未指定，使用默认路径 `./config.yaml`
   - 调用 `runner.config()` 加载配置并初始化组件
6. 启动运行（同步模式）
   - 调用 `runner.run(false)`，阻塞主线程
   - 运行期间：
     - 启动解析器适配器（行情通道）
     - 启动交易适配器（交易通道）
     - 启动交易引擎（CTA/HFT/选股）
     - 等待退出信号
7. 程序退出
   - 返回 0
